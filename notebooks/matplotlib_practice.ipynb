## Matplotlib - Data visualization

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
```

```python
fig, ax = plt.subplots()
ax.scatter(x, y)
```

```python
fig, ax = plt.subplots()
ax.bar(["a", "b", "c"], [3, 7, 5])
```

```python
data = np.random.randn(1000)
fig, ax = plt.subplots()
ax.hist(data, bins=30)
```

```python
fig, ax = plt.subplots()
ax.plot(x, y, label="sin(x)")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Sine Wave")
ax.legend()
```

```python
fig, ax = plt.subplots()
ax.plot(x, np.sin(x), label="sin")
ax.plot(x, np.cos(x), label="cos")
ax.legend()
```

```python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))
axes[0].plot(x, np.sin(x))
axes[0].set_title("sin")
axes[1].plot(x, np.cos(x))
axes[1].set_title("cos")
fig.tight_layout()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(8, 8))
axes[0, 0].plot(x, np.sin(x))
axes[0, 1].plot(x, np.cos(x))
axes[1, 0].hist(data, bins=20)
axes[1, 1].scatter(x[:20], y[:20])
fig.tight_layout()
```

```python
losses = 1 / np.arange(1, 101)

fig, ax = plt.subplots()
ax.plot(losses)
ax.set_yscale("log")
ax.set_xlabel("epoch")
ax.set_ylabel("loss (log scale)")
```

```python
matrix = np.random.randn(5, 5)

fig, ax = plt.subplots()
im = ax.imshow(matrix, cmap="viridis")
fig.colorbar(im, ax=ax)
```

```python
epochs = np.arange(1, 21)
mean_loss = 1 / epochs
std_loss = mean_loss * 0.2

fig, ax = plt.subplots()
ax.plot(epochs, mean_loss, label="mean loss")
ax.fill_between(epochs, mean_loss - std_loss, mean_loss + std_loss, alpha=0.3, label="+-1 std")
ax.legend()
```

```python
fig, ax = plt.subplots()
ax.errorbar(epochs, mean_loss, yerr=std_loss, fmt="-o", capsize=3)
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(
    x, np.sin(x),
    color="tab:blue",
    linestyle="--",
    linewidth=2,
    marker="o",
    markevery=10,
    label="sin",
)
ax.grid(True, alpha=0.3)
ax.legend()
```

```python
fig.savefig("figure.png", dpi=150, bbox_inches="tight")
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
y_true = np.random.randn(100) * 10 + 50   # 100 values centered around 50
y_pred = y_true + np.random.randn(100) * 3  # true value + some noise
```

```python
fig, ax = plt.subplots()
ax.scatter(y_true, y_pred)
lo = min(y_true.min(), y_pred.min())
hi = max(y_true.max(), y_pred.max())
ax.axline([lo, lo], slope=1)
ax.set_xlabel("true values")
ax.set_ylabel("predicted values")
ax.set_title("True vs Predicted values")
```
