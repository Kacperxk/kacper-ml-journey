### Section 2 - Comprehensions and Functional Patterns

```python
result = []
for x in range(10):
    result.append(x ** 2)

result
```

```python
[x ** 2 for x in range(10)]
```

```python
result = []
for x in range(20):
    if x % 2 == 0:
        result.append(x)

result
```

```python
[x for x in range(20) if x % 2 == 0]
```

```python
words = ["hello", "world", "python", "ml"]
result = []
for word in words:
    result.append(word.upper())

result
```

```python
[x.upper() for x in words]
```

```python
result = []
for i in range(3):
    for j in range(3):
        result.append((i, j))

result
```

```python
[(x, y) for x in range(3) for y in range(3)]
```

First loop `for x in` in an outer one

```python
result = []
for i in range(5):
    for j in range(5):
        if i != j:
            result.append(i * j)

result
```

```python
[i * j for i in range(5) for j in range(5) if i != j]
```

```python
def squares(n):
    for x in range(n):
        yield (x + 5)

c = squares(3)
print(next(c))

print("---")

for num in c:
    print(num)
```

```python
random.seed(42)
dataset = [
    ([random.gauss(0, 1) for _ in range(4)], random.randint(0,2)) for _ in range(100)
    ]

dataset[1:3]
```

```python
labels = [label for _, label in dataset]
features = [vector for vector, _ in dataset]
```

```python
result = {0: 0, 1: 0, 2: 0}
for cls in range(3):
    for _, label in dataset:
       if label == cls:
           result[cls] += 1

result

class_counts = {cls: sum(1 for _, label in dataset if label == cls) for cls in range(3)}
class_counts
```

```python
binary_dataset = [(feat, label) for feat, label in dataset if label != 2]
```

```python
mini_batch = dataset[::10]
len(mini_batch)
```

```python
batches = [dataset[i:i+10] for i in range(0, 100, 10)]
len(batches)
flat = [sample for batch in batches for sample in batch]
assert flat == dataset

result = []
for batch in batches:
    for sample in batch:
        result.append(sample)

assert result == dataset
```
