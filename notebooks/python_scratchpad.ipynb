```python
def append_to(element, to=[]):
    to.append(element)
    return to

print(append_to(1))
print(append_to(2))
print(append_to(3))
print(append_to(4, []))
print(append_to(5))
print(append_to(6))
# Python created this default object to=[] once. Now everytime when we leave it empty
# it adds to it since list is mutable.
```
