## Array Creation and dtypes

````python
import numpy as np
````

Exercise 1.1
```
a) A 1D array of integers from 0 to 9
b) A 2D array of zeros with shape (4, 7)
c) A 3D array of ones with shape (2, 3, 5)
d) A 4x4 identity matrix
e) 10 evenly spaced floats between 0 and 1 (inclusive)
f) Integers from 10 down to 0 (inclusive), step -2
g) A 3x3 array filled with the value 99.0
```

````python
# a)
A = np.arange(10)

print(A.shape) # (10,)
print(A.dtype) # int64
print(A.ndim) # 1
print(A.size) # 10
````

````python
# b)
B = np.zeros((4, 7))

print(B.shape) # (4, 7)
print(B.dtype) # float64
print(B.ndim) # 2
print(B.size) # 28
````

````python
# c)
C = np.ones((2, 3, 5))

print(C.shape) # (2, 3, 5)
print(C.dtype) # float64
print(C.ndim) # 3
print(C.size) # 30
````

````python
# d)
D = np.eye(4)

print(D.shape) # (4, 4)
print(D.dtype) # float64
print(D.ndim) # 2
print(D.size) # 16
````

````python
# e)
E = np.linspace(0, 1, 10)

print(E.shape) # (10, )
print(E.dtype) # float64
print(E.ndim) # 1
print(E.size) # 10
````

````python
# f)
F = np.arange(10, -1, -2)

print(F.shape) # (6, )
print(F.dtype) # int64
print(F.ndim) # 1
print(F.size) # 6
````

````python
# g)
G = np.full((3, 3), 99.0)

print(G.shape) # (3, 3)
print(G.dtype) # flfoat64
print(G.ndim) # 2
print(G.size) # 9
````

````python
# Exercise 1.2
a = np.array([1, 2, 3])             # dtype? int64
b = np.array([1.0, 2.0, 3.0])       # dtype? float64
c = np.array([1, 2, 3], dtype=np.float32)  # dtype? float32
d = a + b                            # dtype? (mixing int64 and float64) float64
e = a + c                            # dtype? (mixing int64 and float32) float64
f = np.array([True, False, True])    # dtype? bool
g = f.astype(np.int32)               # dtype? int32
h = np.array([1, 2, 3]).astype(np.float32)  # dtype? float32
````

````python
print(a.dtype)
print(b.dtype)
print(c.dtype)
print(d.dtype)
print(e.dtype)
print(f.dtype)
print(g.dtype)
print(h.dtype)
````
