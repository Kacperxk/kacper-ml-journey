```python
import numpy as np
```

```python
# Exercise 5.1

A = np.array([[1, 2, 3], [4, 5, 6]]) # (2, 3)
B = np.array([[7, 8], [9, 10], [11, 12]]) # (3, 2)
C = np.array([[1, 0, 0], [0, 1, 0], [0, 0, 1]]) # (3, 3) - identity

v = np.array([1, 2, 3]) # (3,)
```

```python
print((A @ B).shape, (A @ B)[0, 0]) # (2, 2) | C[0, 0] == 58
print((A @ C).shape) # (2, 3) - nothing changes, same as multiplying by 1
print((C @ A.T).shape) # (3, 2)
print(v @ A.T) # [14, 32]
print((A.T @ A).shape) # (3, 3) - Gram matrix
print(np.outer(v, v).shape) # (3, 3)
```

```python
# Exercise 5.2

np.random.seed(42)

X = np.random.randn(64, 128)

W1 = np.random.randn(128, 64) * 0.01
b1 = np.zeros(64)

W2 = np.random.randn(64, 32) * 0.01
b2 = np.zeros(32)

W3 = np.random.randn(32, 10) * 0.01
b3 = np.zeros(10)

def relu(x):
    return np.maximum(0, x)

def softmax(logits):
    shifted = logits - logits.max(axis=1, keepdims=True)
    shifted_exp = np.exp(shifted)
    return shifted_exp / shifted_exp.sum(axis=1, keepdims=True)
```

```python
# a)
Z1 = X @ W1 + b1 # (64, 64)
print(Z1.shape)
# b)
A1 = relu(Z1) # (64, 64)
print(A1.shape)
# c
Z2 = A1 @ W2 + b2 # (64, 32)
print(Z2.shape)
# d)
A2 = relu(Z2) # (64, 32)
print(A2.shape)
# e)
Z3 = A2 @ W3 + b3 # (64, 10)
print(Z3.shape)
# f)
probs = softmax(Z3) # (64, 10)
print(probs.shape)
assert np.all(np.isclose(probs.sum(axis=1), 1))
```

```python
# Exercise 5.3

A = np.array([[3, 1, -2], [2, -4, 1], [1, 2, 3]], dtype=float)
b = np.array([5, -1, 10], dtype=float)

# a)
c = np.linalg.solve(A, b)
# b)
assert np.all(np.isclose(A @ c, b))
# c)
A_inv = np.linalg.inv(A)
solution = A_inv @ b
# d)
assert np.all(np.isclose(solution, c))
# e) np.linalg.solve is more stable
```

```python
# Exercise 5.5

np.random.seed(42)
A = np.random.randn(4, 4)
M = A.T @ A

# a)
eigenvalues, eigenvectors = np.linalg.eig(M)
# b)
for i in range(4):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]
    assert np.allclose(M @ v, lam * v)
# c)
inx = np.argsort(eigenvalues)[::-1]
sorted_eigenvalues = eigenvalues[inx]
sorted_eigenvectors = eigenvectors[:, inx]
# d)
assert np.linalg.norm(M @ sorted_eigenvectors[:, 0]) > np.linalg.norm(M @ sorted_eigenvectors[:, -1])
# e)
M_restored = eigenvectors @ np.diag(eigenvalues) @ eigenvectors.T
assert np.allclose(M_restored, M)
```

```python
# Exercise 5.6

np.random.seed(7)
M = np.random.randn(6, 4)

# a)
U, S, Vt = np.linalg.svd(M, full_matrices=False)
# b)
print(U.shape) # left singular vectors
print(S.shape) # singular values, largest to smallest
print(Vt.shape) # right singular vectors
# c)
M_reconstructed = U @ np.diag(S) @ Vt
assert np.allclose(M, M_reconstructed)
# d)
M_approx_2 = U[:, :2] @ np.diag(S[:2]) @ Vt[:2, :]
np.linalg.norm(M - M_approx_2, ord='fro') # ~1,078
# e)
M_approx_1 = U[:, :1] @ np.diag(S[:1]) @ Vt[:1, :]
np.linalg.norm(M - M_approx_1, ord='fro') # ~2,623
# f)
top1 = (S[:1]**2).sum() / (S**2).sum()
top2 = (S[:2]**2).sum() / (S**2).sum()
print(top1.round(4) * 100, top2.round(4) * 100)
```

```python
# Exercise 5.7

np.random.seed(0)
n = 200
x1= np.random.randn(n)
x2 = 3 * x1 + 0.5 * np.random.randn(n)
x3 = np.random.randn(n)
X = np.column_stack([x1, x2, x3])
```

```python
X_centered = X - X.mean(axis=0)
X_cov = np.cov(X_centered.T)
assert X_cov.shape == (3, 3)
eigenvalues, eigenvectors = np.linalg.eig(X_cov)
inx = np.argsort(eigenvalues)[::-1]
sorted_eigenvalues = eigenvalues[inx]
sorted_eigenvectors = eigenvectors[:, inx]
X_2d = X_centered @ sorted_eigenvectors[:, :2]
assert X_2d.shape == (200, 2)
assert X_2d[:, 0].var() > X_2d[:, 1].var()
explained_1 = sorted_eigenvalues[0] / sorted_eigenvalues.sum()
explained_2 = sorted_eigenvalues[1] / sorted_eigenvalues.sum()
explained_3 = sorted_eigenvalues[2] / sorted_eigenvalues.sum()
print(explained_1, explained_2, explained_3)
```
